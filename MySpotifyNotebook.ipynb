```python
# ---------------------------------------------------------
# 4. Unzip the dataset into a folder
# ---------------------------------------------------------
import zipfile

zip_file = "spotify-top-50-playlist-songs-anxods.zip"
extract_folder = "spotify_data"

with zipfile.ZipFile(zip_file, "r") as z:
    z.extractall(extract_folder)

print("Files extracted to:", extract_folder)
```

```python
# ---------------------------------------------------------
# 5. Load the dataset with pandas
# ---------------------------------------------------------
import os
import pandas as pd

extract_folder = "spotify_data"
data_folder = os.path.join(extract_folder, "data")

# List all CSV files inside spotify_data/data/
csv_files = [f for f in os.listdir(data_folder) if f.endswith(".csv")]

csv_files
```

```python
# ---------------------------------------------------------
# 6. Read the CSV (adjust filename if needed)
# ---------------------------------------------------------
df = pd.read_csv(os.path.join(data_folder, "spotify-streaming-top-50-usa.csv"))
df.head()

```

```python
dfs = {}

for file in csv_files:
    path = os.path.join(data_folder, file)
    dfs[file] = pd.read_csv(path)

dfs.keys()
dfs["spotify-streaming-top-50-usa.csv"].head()
```
