```python
# ---------------------------------------------------------
# 4. Unzip the dataset into a folder
# ---------------------------------------------------------
import zipfile

zip_file = "spotify-top-50-playlist-songs-anxods.zip"
extract_folder = "spotify_data"

with zipfile.ZipFile(zip_file, "r") as z:
    z.extractall(extract_folder)

print("Files extracted to:", extract_folder)
```

```python
# ---------------------------------------------------------
# 5. Load the dataset with pandas
# ---------------------------------------------------------
import os
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

%matplotlib inline

extract_folder = "spotify_data"
data_folder = os.path.join(extract_folder, "data")

# List all CSV files inside spotify_data/data/
csv_files = [f for f in os.listdir(data_folder) if f.endswith(".csv")]

csv_files
```

```python
# ---------------------------------------------------------
# 6. Read the CSV (adjust filename if needed)
# ---------------------------------------------------------
df = pd.read_csv(os.path.join(data_folder, "spotify-streaming-top-50-usa.csv"))
df.head()

```

```python
dfs = {}

for file in csv_files:
    path = os.path.join(data_folder, file)
    dfs[file] = pd.read_csv(path)

dfs.keys()
dfs["spotify-streaming-top-50-usa.csv"].head()
```

```python
# By default it gets the top 10 artists

def get_top_artists(df, n=10):
    return (df['artist']
            .str.split(' & ')
            .explode()
            .str.strip()
            .value_counts()
            .head(n))

get_top_artists(dfs["spotify-streaming-top-50-mexico.csv"], 50)
```

```python
top_artists = get_top_artists(dfs["spotify-streaming-top-50-mexico.csv"])

plt.figure(figsize=(10, 6))
plt.barh(top_artists.index, top_artists.values)
plt.title("Top 10 Artists")
plt.xlabel("Number of Songs")
plt.ylabel("Artists")
plt.gca().invert_yaxis() # Invert y-axis to have the highest on top
plt.tight_layout()
plt.show()
```
